```python
# Write a code to reverse a number
'''
to reverse a number we have to do % with 10 whihch will give us last
digit which we want to be first in reveresed number , so we will do rev*10 + lastdigit , this will
do folleing lets say og num = 567
567%10 = 7
rev = 0*10+7 = 7
and then we can elimiate or remove last digit by strict dividing by 10 wich 
will give us 56 (as // give us number without point)
'''
def reverse_number(num) -> bool:
    og :int = num
    reversed_num : int = 0
    while num > 0:
        lastnum = num % 10
        reversed_num  = (reversed_num *10)+lastnum
        num = num//10
    return reversed_num==og

n = int(input())
print(reverse_number(n))
```

```python
# Write the code to find the Fibonacci series upto the nth term.
'''
in fibonacci series teh next number is sum of previous 2 numbers i.e next = current + prev
0,1,1,2,3,5,8,13,21....
'''

def fibo(n)->list:
    series :list = []
    a= 0
    b= 1
    for i in range(2,n):
        c = a+b
        series.append(c)
        a = b
        b = c
    return series


print(fibo(10))
```

```python
# Write code of Greatest Common Divisor 

'''
Greatest common divisor is the largest number which divides both the given number'''

def gcd(n,m)->int:
    x = min(m,n)
    gcd = 0
    for i in range(x-1,0,-1):
        if n%i==0 and m%i==0:
            gcd = max(gcd,i)
    return gcd

print(gcd(36,60))
```

```python
# Write code of  Perfect number
'''
A perfect number is a positive integer that is equal to the sum of 
its proper positive divisors, excluding the number itself. 
For example, 6 is a perfect number because its divisors are 1, 2, and 3, and 1 + 2 + 3 = 6.

'''

def perf_num(n)->bool:
    sum =0
    for i in range(1,n):
        if n%i==0:
            sum = sum+i
    return sum==n

print(perf_num(28))
```

```python
#Check whether two strings are anagrams
''' two strings are anagram if they contain same letter
but in different order, for example "listen" and "silent" are anagrams of each other.

'''

def isanagram(s1,s2)->bool:
    if len(s1)!=len(s2):
        return False
    for s in s1:
        if s not in s2:
            return False
    return True


print(isanagram("listen","silent"))
```

```python
# Check whether a string is a palindrome
'''
a string is palindrome if it is same reversed eg : madam '''

def isPalindrome(s: str)->bool:
    og = s
    rev = ""
    for i in range(len(s)-1,-1,-1):
        rev = rev+s[i]
    return rev==og

print(isPalindrome("madam"))
```

```python
# Frequency of characters in a string
def freq_of_char(s:str)-> dict:
    freq = {}
    for i in s:
        if i in freq:
            freq[i]+=1
        else:
            freq[i]=1
    return freq


d = freq_of_char("hello world")
for k,v in d.items():
    print(f"Character ; {k} and frequency: {v}")
    
```

```python
# Write a code for bubble sort

'''we check adjacent element and swap if they are not in order'''
def bubble_sort(arr:list)->list:
    n = len(arr)
    for i in range(n-1):
        swapped=False
        for j in range(n-1-i):      # as after each iteration last element will be in correct position so we can ignore it
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
                swapped=True
        if not swapped:
            break
    return arr


arr =[]
for _ in range(5):
    arr.append(int(input()))
print(bubble_sort(arr))
```

```python
# Find non-repeating characters in a string

def nrc(s):
    series = {}
    for x in s:
        if x in series:
            series[x]+=1
        else:
            series[x]=1
    lst = [k for k,v in series.items() if v==1 ]
    return lst

print(nrc("hhhelo"))
```

```python
# Write to code to check whether a given year is leap year or not.
'''
a year is a leap year if its divisible by 4 and not divisible by 100 
and its a leap year if its divisible by 400
'''
def is_leapyr(x):
    if x%4 == 0 and x%100!=0:
        print(f'{x} is a leap year')
    elif x%400==0:
        print(f'{x} is a  leap year')
    else:
        print(f'{x} is not a leap year')
```
